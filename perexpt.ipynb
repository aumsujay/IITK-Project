```python
import csv
import os
# Global variables to store data
expenses = []
monthly_budget = 0.0
FILENAME = "expenses.csv"

def add_expense():
    print("\n--- Add New Expense ---")
    date = input("Enter the date (YYYY-MM-DD): ")
    category = input("Enter the category (e.g., Food, Travel): ")
    try:
        amount = float(input("Enter the amount spent: "))
    except ValueError:
        print("Invalid amount. Please enter a number.")
        return
    description = input("Enter a brief description: ")

    expense = {
        'date': date,
        'category': category,
        'amount': amount,
        'description': description
    }
    expenses.append(expense)
    print("Expense added successfully!")

def view_expenses():
    print("\n--- All Expenses ---")
    if not expenses:
        print("No expenses recorded yet.")
        return

    for entry in expenses:
        # Validate data: check if all keys exist and have values
        if all(entry.get(key) for key in ['date', 'category', 'amount', 'description']):
            print(f"{entry['date']} | {entry['category']} | ${entry['amount']:.2f} | {entry['description']}")
        else:
            print("Skipping incomplete entry...")

def set_budget():
    global monthly_budget
    try:
        monthly_budget = float(input("\nEnter your total budget for the month: "))
        print(f"Budget set to ${monthly_budget:.2f}")
    except ValueError:
        print("Invalid input. Please enter a numerical value.")

def track_budget():
    if monthly_budget == 0:
        print("\nPlease set a budget first (Option 3).")
        return
    
    total_spent = sum(item['amount'] for item in expenses)
    print(f"\nTotal spent: ${total_spent:.2f}")
    
    if total_spent > monthly_budget:
        print("Warning: You have exceeded your budget!")
    else:
        remaining = monthly_budget - total_spent
        print(f"You have ${remaining:.2f} left for the month.")

def save_expenses():
    try:
        with open(FILENAME, mode='w', newline='') as file:
            fieldnames = ['date', 'category', 'amount', 'description']
            writer = csv.DictWriter(file, fieldnames=fieldnames)
            writer.writeheader()
            writer.writerows(expenses)
        print("Expenses saved to file.")
    except Exception as e:
        print(f"Error saving file: {e}")

def load_expenses():
    global expenses
    if os.path.exists(FILENAME):
        try:
            with open(FILENAME, mode='r') as file:
                reader = csv.DictReader(file)
                expenses = []
                for row in reader:
                    # Convert amount back to float
                    row['amount'] = float(row['amount'])
                    expenses.append(row)
            print("Previous expenses loaded.")
        except Exception as e:
            print(f"Could not load previous data: {e}")

def main_menu():
    load_expenses()
    while True:
        print("\n--- Expense Tracker Menu ---")
        print("1. Add expense")
        print("2. View expenses")
        print("3. Set and Track budget")
        print("4. Save expenses")
        print("5. Exit")
        
        choice = input("Choose an option (1-5): ")
        
        if choice == '1':
            add_expense()
        elif choice == '2':
            view_expenses()
        elif choice == '3':
            # Sub-menu logic to either set or just view tracking
            sub_choice = input("Press 'S' to Set budget or 'T' to Track: ").upper()
            if sub_choice == 'S':
                set_budget()
            else:
                track_budget()
        elif choice == '4':
            save_expenses()
        elif choice == '5':
            save_expenses()
            print("Goodbye!")
            break
        else:
            print("Invalid choice, please try again.")

if __name__ == "__main__":
    main_menu()
    
```
