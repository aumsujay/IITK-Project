```python
import hashlib
import json
import os
import csv

# --- Utility Functions ---
def hash_password(password):
    return hashlib.sha256(password.encode()).hexdigest()

def load_data(filename):
    if os.path.exists(filename):
        with open(filename, 'r') as f:
            return json.load(f)
    return {}

def save_data(filename, data):
    with open(filename, 'w') as f:
        json.dump(data, f)

# --- Authentication Logic ---
def register():
    users = load_data('users.json')
    username = input("Enter a new username: ")
    if username in users:
        print("Username already exists!")
        return
    password = input("Enter a password: ")
    users[username] = hash_password(password)
    save_data('users.json', users)
    print("Registration successful!")

def login():
    users = load_data('users.json')
    username = input("Username: ")
    password = input("Password: ")
    if username in users and users[username] == hash_password(password):
        print(f"\nWelcome back, {username}!")
        return username
    print("Invalid credentials.")
    return None

# --- Task Management ---
def task_manager(username):
    tasks_file = f"{username}_tasks.json"
    tasks = load_data(tasks_file)

    while True:
        print("\n--- Task Menu ---")
        print("1. Add Task\n2. View Tasks\n3. Complete Task\n4. Delete Task\n5. Logout")
        choice = input("Choice: ")

        if choice == '1':
            desc = input("Task description: ")
            t_id = str(len(tasks) + 1)
            tasks[t_id] = {"desc": desc, "status": "Pending"}
            save_data(tasks_file, tasks)
        elif choice == '2':
            for tid, info in tasks.items():
                print(f"[{tid}] {info['desc']} - {info['status']}")
        elif choice == '3':
            tid = input("Enter Task ID to complete: ")
            if tid in tasks: tasks[tid]['status'] = "Completed"
            save_data(tasks_file, tasks)
        elif choice == '4':
            tid = input("Enter Task ID to delete: ")
            tasks.pop(tid, None)
            save_data(tasks_file, tasks)
        elif choice == '5':
            break

# --- Expense Tracking ---
def expense_tracker(username):
    expenses = []
    budget = 0.0
    filename = f"{username}_expenses.csv"

    # Load existing CSV data
    if os.path.exists(filename):
        with open(filename, 'r') as f:
            reader = csv.DictReader(f)
            expenses = [row for row in reader]

    while True:
        print("\n--- Expense Menu ---")
        print("1. Add Expense\n2. View Expenses\n3. Track Budget\n4. Save & Exit")
        choice = input("Choice: ")

        if choice == '1':
            date = input("Date (YYYY-MM-DD): ")
            cat = input("Category: ")
            amt = float(input("Amount: "))
            desc = input("Description: ")
            expenses.append({"date": date, "category": cat, "amount": amt, "desc": desc})
        elif choice == '2':
            for exp in expenses:
                print(f"{exp['date']} | {exp['category']} | ${exp['amount']} | {exp['desc']}")
        elif choice == '3':
            budget = float(input("Enter monthly budget: "))
            total = sum(float(e['amount']) for e in expenses)
            if total > budget:
                print(f"WARNING: Exceeded by ${total - budget:.2f}!")
            else:
                print(f"Remaining: ${budget - total:.2f}")
        elif choice == '4':
            with open(filename, 'w', newline='') as f:
                writer = csv.DictWriter(f, fieldnames=["date", "category", "amount", "desc"])
                writer.writeheader()
                writer.writerows(expenses)
            break

# --- Main Application Loop ---
def main():
    while True:
        print("\n=== SYSTEM MENU ===\n1. Register\n2. Login\n3. Exit")
        choice = input("Choice: ")
        if choice == '1':
            register()
        elif choice == '2':
            user = login()
            if user:
                print("1. Task Manager\n2. Expense Tracker")
                mode = input("Select Module: ")
                if mode == '1': task_manager(user)
                else: expense_tracker(user)
        elif choice == '3':
            break

if __name__ == "__main__":
    main()
```
